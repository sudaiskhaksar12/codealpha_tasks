## 1. Load the dataset

```python
import pandas as pd

columns = [
    "checking_account",
    "duration",
    "credit_history",
    "purpose",
    "credit_amount",
    "savings_account",
    "employment_since",
    "installment_rate",
    "personal_status_sex",
    "other_debtors",
    "residence_since",
    "property",
    "age",
    "other_installment_plans",
    "housing",
    "existing_credits",
    "job",
    "num_dependents",
    "telephone",
    "foreign_worker",
    "credit_risk"
]

df = pd.read_csv(
    "german.data",
    sep=r"\s+",
    header=None,
    names=columns
)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

### 3. Check data types and missing values

```python
print("Data Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

### 4. Check the target

```python
print("Credit Risk Values:")
print(df["credit_risk"].value_counts())

print("\nCredit Risk Percentages:")
print(df["credit_risk"].value_counts(normalize=True) * 100)
```

```python
print("Statistical Summary of Numerical Features:")
print(df.describe())
```

### 6. Check the categorical features

```python
categorical_columns = df.select_dtypes(include="object").columns

print("Categorical Columns:")
print(categorical_columns)

print("\nNumber of Categories in Each Column:")
for column in categorical_columns:
    print(f"{column}: {df[column].nunique()} categories")
```

### 9. Prepare the target variable

```python

df["target"] = df["credit_risk"].map({
    1: 0,
    2: 1
})

print("Target values:")
print(df["target"].value_counts())

print("\nTarget percentages:")
print(df["target"].value_counts(normalize=True) * 100)
```

### 10. Separate Features (X) and Target (y)

```python

X = df.drop(columns=["credit_risk", "target"])
y = df["target"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeature columns:")
print(X.columns.tolist())
```

### 11. Split the Data

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting data:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

### 12. Preprocess the data

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

# Identify numerical and categorical columns
numerical_columns = X.select_dtypes(include="number").columns.tolist()
categorical_columns = X.select_dtypes(include="str").columns.tolist()

print("Numerical columns:", len(numerical_columns))
print("Categorical columns:", len(categorical_columns))

# Preprocess numerical features
numerical_transformer = StandardScaler()

# Preprocess categorical features
categorical_transformer = OneHotEncoder(
    handle_unknown="ignore"
)

# Combine both preprocessing steps
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_transformer, numerical_columns),
        ("cat", categorical_transformer, categorical_columns)
    ]
)

print("\nPreprocessor created successfully!")
```

### 13. Train the Logistic Regression model

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Create the complete model pipeline
logistic_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=2000,
        random_state=42
    ))
])

# Train the model
logistic_model.fit(X_train, y_train)

# Make predictions on unseen test data
y_pred_lr = logistic_model.predict(X_test)

# Evaluate the model
print("Logistic Regression Results")
print("-" * 35)

print("Accuracy:", round(accuracy_score(y_test, y_pred_lr), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_lr,
    labels=[0, 1],
    target_names=["Good Credit", "Bad Credit"],
    zero_division=0
))
```

### 14. Train a Decision Tree model

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report

# Create the Decision Tree pipeline
decision_tree_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", DecisionTreeClassifier(
        max_depth=5,
        class_weight="balanced",
        random_state=42
    ))
])

# Train the model
decision_tree_model.fit(X_train, y_train)

# Predict on test data
y_pred_dt = decision_tree_model.predict(X_test)

# Evaluate the model
print("Decision Tree Results")
print("-" * 35)

print("Accuracy:", round(accuracy_score(y_test, y_pred_dt), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_dt,
    labels=[0, 1],
    target_names=["Good Credit", "Bad Credit"],
    zero_division=0
))
```

### 15. Random Forest Model 

```python

from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report

# Create the tuned Random Forest pipeline
rf_tuned_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=300,
        max_depth=12,
        min_samples_leaf=2,
        class_weight={0: 1, 1: 2},
        random_state=42,
        n_jobs=-1
    ))
])

# Train the model
rf_tuned_model.fit(X_train, y_train)

# Make predictions
y_pred_rf_tuned = rf_tuned_model.predict(X_test)

# Evaluate the model
print("Tuned Random Forest Results")
print("-" * 35)

print("Accuracy:", round(
    accuracy_score(y_test, y_pred_rf_tuned), 4
))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_rf_tuned,
    labels=[0, 1],
    target_names=["Good Credit", "Bad Credit"],
    zero_division=0
))
```

### 16. Import and Train The SVM model

```python
from sklearn.svm import SVC

print("SVM imported successfully!")
```

```python
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report

svm_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", SVC(class_weight="balanced", random_state=42))
])

svm_model.fit(X_train, y_train)

y_pred_svm = svm_model.predict(X_test)

print("SVM Accuracy:", round(accuracy_score(y_test, y_pred_svm), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_svm,
    labels=[0, 1],
    target_names=["Good Credit", "Bad Credit"],
    zero_division=0
))
```

### 17. Compare Models

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
import pandas as pd

results = []

models = {
    "Logistic Regression": y_pred_lr,
    "Decision Tree": y_pred_dt,
    "SVM": y_pred_svm,
    "Random Forest": y_pred_rf_tuned
}

for name, predictions in models.items():
    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Bad Credit Precision": precision_score(
            y_test, predictions, pos_label=1, zero_division=0
        ),
        "Bad Credit Recall": recall_score(
            y_test, predictions, pos_label=1, zero_division=0
        ),
        "Bad Credit F1-Score": f1_score(
            y_test, predictions, pos_label=1, zero_division=0
        )
    })

comparison_df = pd.DataFrame(results)

print("Credit Scoring Model Comparison")
print("-" * 75)
print(comparison_df.round(3).to_string(index=False))
```

### 18. All four confusion matrices

```python

import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Predictions from all four models
models = {
    "Logistic Regression": y_pred_lr,
    "Decision Tree": y_pred_dt,
    "SVM": y_pred_svm,
    "Tuned Random Forest": y_pred_rf_tuned
}

# Display a confusion matrix for each model
for name, predictions in models.items():
    cm = confusion_matrix(y_test, predictions, labels=[0, 1])

    display = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["Good Credit", "Bad Credit"]
    )

    display.plot()
    plt.title(f"{name} - Confusion Matrix")
    plt.tight_layout()
    plt.show()
```

## Model Comparison and Conclusion

Four classification models were evaluated for the German Credit Scoring dataset: Logistic Regression, Decision Tree, Support Vector Machine (SVM), and Random Forest.

### Results

* **Logistic Regression** achieved the highest accuracy of 78%.
* **SVM** achieved the highest bad-credit recall of 76.7% and the highest bad-credit F1-score of 0.657.
* **Decision Tree** achieved a bad-credit recall of 66.7%, but its overall accuracy was lower at 57%.
* **Random Forest** achieved 77% accuracy, but its bad-credit recall was 50%.

### Conclusion

Based on the reported results, SVM is the strongest candidate when identifying bad-credit cases is a priority. Logistic Regression achieved the highest overall accuracy.

The final model should be selected after validating the results using cross-validation and checking the costs of false predictions. These results are educational and should not be used alone to make real lending decisions.
